# 02 — Independent validation

The quick suite checks independent discrete operators, exact free packets, dense exponential references, closed reversal and relative phase. The **full** suite adds temporal/spatial refinement, oscillator moments, domain/CAP studies, scattering, eigenstates and geometry.

This notebook deliberately runs the quick suite so revision stays fast. Its report lists full-suite checks that were not run; a quick pass is not a full-suite pass.

```python
from pathlib import Path
import json
import sys
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from IPython.display import Image, display

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src" / "wavepacket").is_dir())
print("Python:", sys.executable)

OUT = ROOT / "results" / "notebook_runs" / "02_validation"
OUT.mkdir(parents=True, exist_ok=True)
from wavepacket.validation import run_validation

report = run_validation("quick", OUT)
assert report["status"] == "passed", report["summary"]
print(report["summary"])
for case in report["cases"]:
    print(case["id"], case["status"], case["metrics"])
print("Not run by this notebook:", report["not_run"])
```

```python
fig, ax = plt.subplots(figsize=(9, 3.5))
names = [case["id"] for case in report["cases"]]
ax.barh(names, [case["elapsed_seconds"] for case in report["cases"]])
ax.set(xlabel="elapsed time (s)", title="Measured quick-suite execution; not a solver performance comparison")
ax.invert_yaxis()
fig.tight_layout()
fig.savefig(OUT / "validation_cases.png", dpi=150)
plt.close(fig)
display(Image(filename=str(OUT / "validation_cases.png")))
```
